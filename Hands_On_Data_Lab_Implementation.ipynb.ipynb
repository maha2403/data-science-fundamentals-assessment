```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    "Name": ["Arun", "Priya", "Kavin", "Divya", "Ravi", "Meena"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "Sales"],
    "Salary": [25000, 30000, 28000, 35000, 27000, 40000],
    "Experience": [1, 3, 2, 5, 2, 6]
}

df = pd.DataFrame(data)

print("Sample Employee Dataset:")
display(df)
```

```python

# 1. Display the first 5 rows
print("First 5 rows:")
display(df.head())

# 2. Check dataset information
print("\nDataset Information:")
df.info()

# 3. Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 4. Calculate average salary
print("\nAverage Salary:")
print(df["Salary"].mean())

# 5. Find the highest salary
print("\nHighest Salary:")
print(df["Salary"].max())

# 6. Sort employees by salary
print("\nEmployees sorted by salary:")
display(df.sort_values(by="Salary", ascending=False))
```

```python

# Department-wise employee count
print("Employees in each department:")
print(df["Department"].value_counts())

# Average salary by department
avg_salary = df.groupby("Department")["Salary"].mean()

print("\nAverage Salary by Department:")
print(avg_salary)

# Bar chart: Average salary by department
avg_salary.plot(kind="bar", figsize=(8, 5))

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Create a dataset with missing values
messy_data = {
    "Name": ["Arun", "Priya", "Kavin", "Divya", "Ravi"],
    "Age": [22, 24, np.nan, 23, 25],
    "Salary": [25000, 30000, 28000, np.nan, 35000]
}

clean_df = pd.DataFrame(messy_data)

print("Dataset before cleaning:")
display(clean_df)

# Check missing values
print("Missing values before cleaning:")
print(clean_df.isnull().sum())

# Fill missing values with column averages
clean_df["Age"] = clean_df["Age"].fillna(clean_df["Age"].mean())
clean_df["Salary"] = clean_df["Salary"].fillna(clean_df["Salary"].mean())

print("\nDataset after cleaning:")
display(clean_df)

print("Missing values after cleaning:")
print(clean_df.isnull().sum())
```
